1.Which Bangalore locations have the highest average restaurant ratings?


```python
conn = sqlite3.connect('restaurant1.db')
query1 = """
SELECT
    location,
    ROUND(AVG(rate), 2) AS avg_rating,
    COUNT(*) AS total_restaurants,
    ROUND(AVG( approx_cost_for_two), 0) AS avg_cost
FROM restaurants1
WHERE rate IS NOT NULL
GROUP BY location
HAVING total_restaurants >=30
ORDER BY avg_rating DESC
LIMIT 10;
"""
premium_locations_df = pd.read_sql_query(query1, conn)
table = tabulate(premium_locations_df
    ,
    headers='keys',
    tablefmt='pipe'
)
print(table)
```

5.What price range delivers the best customer satisfaction?


```python
query5 = """
SELECT
    CASE
        WHEN approx_cost_for_two <= 500 THEN 'Budget (0-500)'
        WHEN approx_cost_for_two <= 1000 THEN 'Mid-Range (501-1000)'
        WHEN approx_cost_for_two <= 2000 THEN 'Premium (1001-2000)'
        ELSE 'Luxury (2000+)'
    END AS price_segment,
    ROUND(AVG(rate), 2) AS avg_rating,
    COUNT(*) AS restaurant_count,
    ROUND(AVG(votes), 0) AS avg_engagement
FROM restaurants1
WHERE rate IS NOT NULL AND approx_cost_for_two IS NOT NULL
GROUP BY price_segment
ORDER BY avg_rating DESC;
"""
pricing_strategy_df = pd.read_sql_query(query5, conn)

table = tabulate(pricing_strategy_df
    ,
    headers='keys',
    tablefmt='grid'
)
print(table)
```


15.Which restaurants are top performers within each pricing segment?


```python
query15 = """
WITH PriceSegments AS (
    SELECT
        restaurant_name,
        location,
        cuisines,
        rate,
        votes,
        approx_cost_for_two,
        CASE
            WHEN approx_cost_for_two <= 500 THEN 'Budget'
            WHEN approx_cost_for_two <= 1000 THEN 'Mid-Range'
            WHEN approx_cost_for_two <= 2000 THEN 'Premium'
            ELSE 'Luxury'
        END AS price_segment
    FROM restaurants1
    WHERE rate IS NOT NULL
)
SELECT * FROM (
    SELECT *,
           RANK() OVER (PARTITION BY price_segment ORDER BY rate DESC, votes DESC) as rank
    FROM PriceSegments
)
WHERE rank <= 3;
"""
benchmark_partners_df = pd.read_sql_query(query15, conn)
table = tabulate(benchmark_partners_df
    ,
    headers='keys',
    tablefmt='grid'
)
print(table)

```

10. What is the volatility (Standard Deviation) of daily sales?


```python
query10 = """
SELECT
    AVG(daily_total) AS avg_daily_revenue,
    -- Standard Deviation formula (Population)
    SQRT(AVG(daily_total * daily_total) - AVG(daily_total) * AVG(daily_total)) AS daily_volatility
FROM (
    SELECT order_date, SUM(order_value) AS daily_total
    FROM orders
    GROUP BY order_date
);
"""
df10 = pd.read_sql_query(query10, conn)
print(tabulate(df10, headers='keys', tablefmt='pipe', showindex=False))
```

```python
%%writefile app.py
import streamlit as st
import sqlite3
import pandas as pd
import os

st.set_page_config(page_title="Uber Eats & Order Analytics", layout="wide")

# Use ONE database name for everything
DB_NAME = 'restaurant1.db'

@st.cache_data
def load_and_prep_all_data():
    # 1. Load Restaurant Data
    try:
        # Update these paths to where your files actually sit
        df_rest = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/restaurantdata (1).csv')
        df_rest.columns = [c.replace(' ', '_').lower() for c in df_rest.columns]

        # 2. Load Order Data
        df_order = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/cleaned order_data (1).csv')
        df_order['order_date'] = pd.to_datetime(df_order['order_date'])

        # 3. Save both to the SAME database
        conn = sqlite3.connect('restaurant1.db')
        df_rest.to_sql('restaurants1', conn, index=False, if_exists='replace')
        df_order.to_sql('orders', conn, index=False, if_exists='replace')
        conn.close()

        return df_order
    except Exception as e:
        st.error(f"Error loading data: {e}")
        return pd.DataFrame()

df_orders = load_and_prep_all_data()

# Helper for SQL
def run_sql(query, params=None):
    with sqlite3.connect('restaurant1.db') as conn:
        if params:
            return pd.read_sql_query(query, conn, params=params)
        return pd.read_sql_query(query, conn)

# 3. Sidebar Navigation
st.sidebar.title("🚀 Navigation")
page = st.sidebar.radio("Select Dashboard",
    ["Restaurant Dashboard", "Restaurant Q&A", "Order Dashboard", "Order BI (SQL Queries)"])

# --- PAGE 1: RESTAURANT DASHBOARD ---
if page == "Restaurant Dashboard":
    st.title("🍽️ Bangalore Restaurant Analysis")
    st.sidebar.header("Filters")

    try:
        loc_df = run_sql("SELECT DISTINCT location FROM restaurants1 WHERE location IS NOT NULL")
        locations = ["All"] + sorted(loc_df['location'].tolist())

        selected_loc = st.sidebar.selectbox("Select Location", locations)
        order_online = st.sidebar.radio("Online Order Available?", ["Any", "Yes", "No"])
        max_price = st.sidebar.slider("Max Cost for Two", 0, 5000, 5000)

        # Build Query
        query = "SELECT * FROM restaurants1 WHERE approx_cost_for_two <= ?"
        params = [max_price]

        if selected_loc != "All":
            query += " AND location = ?"
            params.append(selected_loc)

        if order_online != "Any":
            query += " AND online_order = ?"
            params.append(order_online)

        # Execute
        conn = sqlite3.connect('restaurant1.db')
        filtered_df = pd.read_sql_query(query, conn, params=params)
        conn.close()

        if not filtered_df.empty:
            col1, col2, col3 = st.columns(3)
            col1.metric("Total Found", len(filtered_df))
            col2.metric("Avg Rating", round(filtered_df['rate'].mean(), 2))
            col3.metric("Avg Cost", f"₹{int(filtered_df['approx_cost_for_two'].mean())}")

            st.dataframe(filtered_df, use_container_width=True, hide_index=True)
        else:
            st.warning("No restaurants match your filters.")

    except Exception as e:
        st.error(f"Error loading dashboard: {e}")


# --- PAGE 2: RESTAURANT Q&A ---
elif page == "Restaurant Q&A":
    st.title("🔍 Restaurant Business Intelligence")
    q_rest = {
         "1. Highest Rated Locations": "SELECT location, AVG(rate) as avg_rating FROM restaurants1 GROUP BY location ORDER BY avg_rating DESC LIMIT 10",
        "2. Over-saturated Locations": "SELECT location, COUNT(*) as total_restaurants FROM restaurants1 GROUP BY location ORDER BY total_restaurants DESC LIMIT 10",
        "3. Online Order vs Ratings": "SELECT online_order, AVG(rate) as avg_rating FROM restaurants1 GROUP BY online_order",
        "4. Table Booking vs Ratings": "SELECT book_table, AVG(rate) as avg_rating FROM restaurants1 GROUP BY book_table",
        "5. Best Satisfaction Price Range": "SELECT approx_cost_for_two, AVG(rate) as avg_rating FROM restaurants1 GROUP BY approx_cost_for_two ORDER BY avg_rating DESC LIMIT 10",
        "6. Performance by Price Segment": "SELECT CASE WHEN approx_cost_for_two < 500 THEN 'Low' WHEN approx_cost_for_two BETWEEN 500 AND 1500 THEN 'Mid' ELSE 'Premium' END as segment, AVG(rate) as avg_rating FROM restaurants1 GROUP BY segment",
        "7. Most Common Cuisines": "SELECT cuisines, COUNT(*) as count FROM restaurants1 GROUP BY cuisines ORDER BY count DESC LIMIT 10",
        "8. Highest Rated Cuisines": "SELECT cuisines, AVG(rate) as avg_rating FROM restaurants1 GROUP BY cuisines HAVING COUNT(*) > 10 ORDER BY avg_rating DESC LIMIT 10",
        "9. Niche High-Performers": "SELECT cuisines, COUNT(*) as count, AVG(rate) as avg_rating FROM restaurants1 GROUP BY cuisines HAVING count < 5 ORDER BY avg_rating DESC LIMIT 10",
        "10. Cost vs Rating Relationship": "SELECT approx_cost_for_two, AVG(rate) as avg_rating FROM restaurants1 GROUP BY approx_cost_for_two ORDER BY approx_cost_for_two ASC",
        "11. Ideal Locations for Premium Onboarding": "SELECT location, AVG(approx_cost_for_two) as avg_cost, AVG(rate) as avg_rating FROM restaurants1 GROUP BY location HAVING avg_cost > 1000 ORDER BY avg_rating DESC",
        "12. High Demand / Low Rating (Opportunity)": "SELECT location, COUNT(*) as total_orders, AVG(rate) as avg_rating FROM restaurants1 GROUP BY location HAVING total_orders > 100 ORDER BY avg_rating ASC LIMIT 10",
        "13. Dual Service Performance": "SELECT online_order, book_table, AVG(rate) as avg_rating FROM restaurants1 GROUP BY online_order, book_table",
        "14. Success Factors for Delivery": "SELECT online_order, AVG(rate) as avg_rating, SUM(votes) as total_votes FROM restaurants1 GROUP BY online_order",
        "15. Top Performers by Segment": "SELECT name, approx_cost_for_two, rate, location FROM restaurants1 WHERE rate > 4.5 ORDER BY approx_cost_for_two DESC LIMIT 20"
    }
    sel = st.selectbox("Question:", list(q_rest.keys()))
    st.dataframe(run_sql(q_rest[sel]), use_container_width=True)

# --- PAGE 3: ORDER DASHBOARD (Visuals) ---
elif page == "Order Dashboard":
    st.title("Order Dashboard")
    st.sidebar.header("Filters")
    rest_list = df_orders['restaurant_name'].unique()
    selected_res = st.sidebar.multiselect("Restaurants", rest_list, default=rest_list[:5])

    f_df = df_orders[df_orders['restaurant_name'].isin(selected_res)]

    col1, col2 = st.columns(2)
    col1.metric("Total Revenue", f"₹{f_df['order_value'].sum():,.2f}")
    col2.metric("Avg Order", f"₹{f_df['order_value'].mean():.2f}")

    st.subheader("Revenue by Restaurant")
    st.bar_chart(f_df.groupby('restaurant_name')['order_value'].sum())

# --- PAGE 4: ORDER BI (New SQL Queries) ---
elif page == "Order BI (SQL Queries)":
    st.title("💡 Order Insights via SQL")

    queries = {
        "1. Total Revenue": "SELECT SUM(order_value) as result FROM orders",
        "2. Discount vs AOV": "SELECT discount_used, AVG(order_value) as result FROM orders GROUP BY discount_used",
        "3. Most Frequent Payment": "SELECT payment_method, COUNT(*) as result FROM orders GROUP BY payment_method ORDER BY result DESC",
        "4. Monthly Spikes": "SELECT strftime('%m', order_date) as month, COUNT(*) as result FROM orders GROUP BY month ORDER BY month",
        "5. Top Brand by Volume": "SELECT restaurant_name, SUM(order_value) as result FROM orders GROUP BY restaurant_name ORDER BY result DESC LIMIT 10",
        "6. Typical Spend (Avg)": "SELECT AVG(order_value) as result FROM orders",
        "7. Discount Usage %": "SELECT (SUM(CASE WHEN discount_used = 'Yes' THEN 1 ELSE 0 END) * 100.0 / COUNT(*)) as result FROM orders",
        "8. Pareto Analysis (Top 5% Revenue)": "SELECT restaurant_name, SUM(order_value) as rev FROM orders GROUP BY restaurant_name ORDER BY rev DESC",
        "9. Card vs Cash AOV": "SELECT payment_method, AVG(order_value) as result FROM orders WHERE payment_method IN ('Card', 'Cash') GROUP BY payment_method",
        "10. Daily Sales Std Dev": """SELECT AVG(daily_sum) as avg_sales,
            (SUM((daily_sum - (SELECT AVG(d_sum) FROM (SELECT SUM(order_value) as d_sum FROM orders GROUP BY order_date))) *
            (daily_sum - (SELECT AVG(d_sum) FROM (SELECT SUM(order_value) as d_sum FROM orders GROUP BY order_date)))) / COUNT(*)) as variance
            FROM (SELECT SUM(order_value) as daily_sum FROM orders GROUP BY order_date)"""
    }
    sel = st.selectbox("Question:", list(queries.keys()))
    st.dataframe(run_sql(queries[sel]), use_container_width=True)
```
